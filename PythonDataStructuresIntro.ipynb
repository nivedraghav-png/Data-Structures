**Python Data Structures**

**1. List**


```python
Numbers = [10,25,13,99,47]     #Create a List of 5 Random Numbers     
print(Numbers)                
```

```python
Numbers = [10,25,13,99,47]   

Numbers.extend([33,58,84])     #Inserting 3 New Values 

print(Numbers)
```

```python
Numbers = [10,25,13,99,47]

for n in Numbers:             #print each element using for loop
    print(n)
```

**2. Dictionary**

```python
Dict = {'name':'John', 'age':'25', 'address':'New York'}     #create a Dictionary

print(Dict)
```

```python
Dict['phone'] = '1234567890'     #Add Phone to the Dictionary

print(Dict)
```

**3. Set**

```python
numbers = {1,2,3,4,5}          #create a set

print(numbers)
```

```python
numbers.add(6)                #Add Value 6 to the set

print(numbers)
```

```python
numbers.remove(3)            #Remove value 3 from the set

print(numbers)
```

**4. Tuple**

```python
T_numbers = (1,2,3,4)         #create a Tuple

T_numbers
```

```python
print(len(T_numbers))        #Length of a Tuple
```
